```python
import control
import numpy as np
import numba as nb
```

```python
@nb.njit()
def state_update(A, B, C, D, x, u, dt):
	dx = np.dot(A, x) + np.dot(B, u)
	x_next = x + dx * dt       # Euler integration
	out = np.dot(C, x_next) + np.dot(D , u)  # System output
	return x_next, out, dx
```
